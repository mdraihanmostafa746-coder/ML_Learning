## KNN (K Nearest Value)

The K-Nearest Neighbors (KNN) algorithm works by assuming that similar data points exist in close proximity to each other, using the core principle of "birds of a feather flock together". Instead of building a complex mathematical model during a training phase, KNN simply memorizes the entire training dataset and defers all computations until a prediction for a new data point is needed. Because of this behavior, it is officially classified as a lazy learner and a non-parametric algorithm.
