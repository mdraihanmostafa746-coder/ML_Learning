## Decision tree

A decision tree is a supervised machine learning algorithm used for both classification (predicting a category) and regression (predicting a continuous number). It is highly popular because it operates like a flowchart, making its reasoning process incredibly easy for humans to read, visualize, and interpret

               [ Root Node ]
             Is Age > 30 years?
               /          \
        Yes   /            \   No
             /              \
     [Decision Node]      [ Leaf Node ]
   Is Income > $70,000?   Will NOT Buy
      /          \
Yes  /            \  No
    /              \
[ Leaf Node ]   [ Leaf Node ]
  Will Buy      Will NOT Buy


```python
from sklearn.tree import DecisionTreeClassifier

X =[
    [7,2], #Apple
    [8,3],#Apple
    [9,8],#Orange
    [10,9]#Orange
]

y = [0, 0, 1, 1] #0 = Apple & 1 = Orrange

model = DecisionTreeClassifier()

model.fit(X,y)

size = float(input("Enter the fruit size in cm L: "))

shade = float(input("Enter the fruit color shade b/w (1 - 10) : "))

result = model.predict([[size, shade]]) [0]

if result == 0:
    print("This is likely an Apple.")

else:
    print("This is likely an Orange.")
```
