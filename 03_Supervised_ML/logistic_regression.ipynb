## Logistic regression 

Logistic regression is a supervised machine learning algorithm used to predict the probability of a categorical outcome. Despite having "regression" in its name, it is primarily used for classification tasks—most commonly binary classification, where the output falls into one of two categories (such as Yes/No, Spam/Not Spam, or 0/1)

```python
from sklearn.linear_model import LogisticRegression

X = [[1], [2], [3], [4], [5]]
y = [0, 0, 1, 1, 1]

model = LogisticRegression()

model.fit(X, y)

hours = float (input("Enter how much hours you studies : "))

result = model.predict([[hours]]) [0]

if result == 1:
    print(f" Based on your hours {hours} you are likely to PASS ")
else :
    print(f" Based on your hours{hours} you are likely to FAIL")
```
